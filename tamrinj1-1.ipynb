```python
import numpy as np 
import matplotlib.pyplot as plt
from scipy import stats
```

```python
x=[23,33,12,35,53]
y=[12,43,12,34,31]
plt.scatter(x,y)
plt.show()
```

```python
slope,intercept,r,p,std_err=stats.linregress(x,y)
def myfunc(x):
    return slope*x+intercept
mymodel=list(map(myfunc,x))
plt.scatter(x,y)
plt.plot(x,mymodel,"r-")
plt.show()
```
